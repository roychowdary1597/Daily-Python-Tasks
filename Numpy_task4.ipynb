## 7. Create a Custom Grading System
    Generate an array of 10 random marks between 0 and 100. Use np.where() to assign grades:
    'A' for marks >= 75
    'B' for 50–74
    'C' for below 50
    Hint: Use np.select() or chain np.where()
    
    Objective: 
    
    Apply broadcasting and conditional logic to real-world-style data


```python
marks = np.random.randint(0,100,10)
```

```python
marks
```

```python
assign1 = np.where(marks >= 75,'A',(
          np.where(marks >= 50,'B','C')))
```

```python
print('Marks:',marks)
print('Grades:',assign1)
```

```python
using np.select
```

```python
condition = [marks >=75,(marks >=50)&(marks <75),marks<50]
```

```python
choice = ['A','B','C']
```

```python
assign2 = np.select(condition,choice)
```

```python
print('Marks:',marks)
```

```python
print('Grades:',assign2)
```
