## 10. Array Iteration
    Create a 2D NumPy array of shape (2, 3) with any integers.
    Iterate through the array using a basic for loop.
    Then, iterate using np.nditer().
    
    Objective: 
    
    Explore different ways of traversing NumPy arrays.


```python
arr2d = np.array([[1,2],[3,4],[5,6]])
```

```python
arr2d
```

```python
for i in arr2d:
    for j in i:
        print(j)
```

```python
for i in np.nditer(arr2d):
    print(i)

```

## 11. Operations on Random Float Arrays
    Generate two 1D arrays (length 5 each) with random float numbers between 1 and 10.

    Apply mathematical functions: np.sqrt(), np.exp(), np.sin(), np.log().
    
    Objective: 
    
    Get hands-on with math operations and functions in NumPy.



```python
arr1 = np.random.uniform(1,10,5)
arr2 = np.random.uniform(1,10,5)
```

```python
print(arr1)
print(arr2)
```

```python
#element wise operations
add = arr1 + arr2
print(add)
```

```python
sub = arr1 - arr2
print(sub)
```

```python
mul = arr1 * arr2
print(mul)
```

```python
div = arr1 / arr2
print(div)
```

```python
mod = arr1 % arr2
print(mod)
```

```python
floor_div = arr1 // arr2
print(floor_div)
```

```python
sqrt_arr1 = np.sqrt(arr1)
print(sqrt_arr1)
```

```python
exp_arr1 = np.exp(arr1)
print
sin_arr1 = np.sin(arr1)
log_arr1 = np.log(arr1)
```
