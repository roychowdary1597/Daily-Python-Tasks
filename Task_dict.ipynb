## 1: Restaurant Menu Storage
>- Create a dictionary where dish names serve as keys and their prices as values. Display the menu in a user-friendly format.
>- Input:Dishes = ["Pasta", "Biryani", "Apricot"]
>- Prices = [200, 300, 180]
>- Expected Output:{“Pasta” : 200, ”Biryani” : 300, “Apricot” : 180}


```python
dishes = ["Pasta", "Biryani", "Apricot"]
```

```python
prices = [200,300,180]
```

```python
menu = dict(zip(dishes,prices))
```

```python
print(menu)
```

## 4: Employee Salary Management
>- Create a dictionary where employee names are keys and salaries are values. Identify the employee with the highest salary.
>- >- Highest Salary: Pavan - ₹75000



```python
salaries = {
    "Ved": 60000,
    "Pavan": 75000,
    "Teja": 50000
}

max_salary_employee = max(salaries, key=salaries.get)
max_salary = salaries[max_salary_employee]

print(f"Highest Salary: {max_salary_employee} - ₹{max_salary}")
```

## 6: Word Frequency Counter
>-Given a paragraph of text, count the frequency of each unique word and store it in a dictionary.


```python
text = "apple banana apple orange banana apple"
```

```python
words = text.split()
print(words)
```

```python
fruits = {}
for fruit in words:
    fruits[fruit] = fruits.get(fruit,0)+1
print(fruits)
```

## 7: Customer Order Management
>- Manage customer orders using a dictionary where the key is the order number, and the value is a list of ordered items.


```python
orders = {
    "1001": ["Burger", "Fries"],
    "1002": ["Pizza", "Soda"]
}
```

```python
order_id = '1001'
```

```python
if order_id in orders:
    items = orders[order_id]
    cnt_items = len(items)
    list_item = ' and '.join(items)
    print(f'{order_id} contains {cnt_items} items,they are {list_item}')
else:
    print(f'Order ID {order_id} not found')
```
