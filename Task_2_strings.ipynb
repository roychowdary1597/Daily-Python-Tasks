## 1)Mask a credit card number (show only the last 4 digits).

>- Input: "1234567812345678"

>- Output: "************5678"


```python
creditCard = '1234567812345678'
```

```python
maskCard = "*" * (len(creditCard) - 4) + creditCard[12: ]
```

```python
print(maskCard)
```
