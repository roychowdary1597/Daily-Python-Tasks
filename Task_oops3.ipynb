## 1. Create a Person class with private attributes __name and __age. Implement getter and setter methods for accessing and modifying these attributes.	
    Modify the Person class from the above problem to use properties for accessing and modifying the private attributes.
    Objective:
    Understand how to implement encapsulation in Python using private attributes and manage access through getter and setter methods with property decorators (@property) to simplify attribute access and modification.
    Sample Input:
    p1 = Person("Johnny", 30)
    
    print("Name:", p1.name)
    print("Age:", p1.age)
    
    p1.name = "Arvind"
    p1.age = 28
    
    print("Updated Name:", p1.name)
    print("Updated Age:", p1.age)
    
    Expected Output:
    Name: Johnny
    Age: 30
    Updated Name: Arvind
    Updated Age: 28


```python
class Person:
    def __init__(self,name,age):
        self.__name = name
        self.__age = age
    @property
    def name(self):
        return self.__name

    @name.setter
    def name(self,value):
        self.__name = value

    @property
    def age(self):
        return self.__age

    @age.setter
    def age(self,value):
        self.__age = value
```

```python
p1 = Person("Johnny", 30)

print("Name:", p1.name)
print("Age:", p1.age)

p1.name = "Arvind"
p1.age = 28

print("Updated Name:", p1.name)
print("Updated Age:", p1.age)
```

## 2. Create a BankAccount class with a protected attribute _balance. Implement a method get_balance() to retrieve the balance.
    Create a Python module calculator with a private function _ _add that adds two numbers. Implement a public function add_numbers that calls the private function.
    Objective:
     Learn the concept of access specifiers in Python — specifically protected (_) and private (__) attributes and methods. Understand how to use protected attributes in classes and how to create private functions inside a module while exposing access through public wrapper functions.
    Sample Input:
    # For BankAccount class
    account = BankAccount(5000)
    print("Balance:", account.get_balance())
    
    # For calculator module
    from calculator import add_numbers
    print("Addition Result:", add_numbers(10, 15))
    
    
    Example output- Car:
    Balance: 5000
    Addition Result: 25
    


```python
class BankAccount:
    def __init__(self,balance):
        self._balance = balance

    def get_balance(self):
        return self._balance
```

```python
account = BankAccount(5000)
print("Balance:", account.get_balance())
```

```python
class Calculator:
    def __init__(self):
        pass

    def __add(self, a, b): 
        return a + b

    def add_numbers(self, a, b): 
        return self.__add(a, b)
```

```python
calc = Calculator()
result = calc.add_numbers(10, 15)
print("Addition Result:", result)
```

## 3. Create an abstract class Shape with an abstract method calculate_area(). Define two subclasses, Rectangle and Circle, that inherit from Shape and implement the calculate_area() method.
    Objective:
    Understand how to implement abstraction in Python using the abc (Abstract Base Class) module. Learn how to define and enforce a common interface (calculate_area()) in subclasses through the use of an abstract base class Shape.
    Sample Input:
    rect = Rectangle(10, 5)
    circle = Circle(7)
    
    print("Area of Rectangle:", rect.calculate_area())
    print("Area of Circle:", circle.calculate_area())
    
    Expected Output:
    Area of Rectangle: 50
    Area of Circle: 153.93804002589985


```python
from abc import ABC, abstractmethod
import math


class Shape(ABC):
    @abstractmethod
    def calculate_area(self):
        pass


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def calculate_area(self):
        return self.length * self.width


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return math.pi * self.radius ** 2
```

```python
rect = Rectangle(10, 5)
circle = Circle(7)

print("Area of Rectangle:", rect.calculate_area())
print("Area of Circle:", circle.calculate_area())
```
