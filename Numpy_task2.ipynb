## 2. Generate Arrays of Different Dimensions.
    Generate five NumPy arrays using np.arange():
    0D array (a single value)
    1D array (e.g., 0 to 4)
    2D array (reshape a 1D array into 2x3)
    3D array (create a 2x2x2 cube)
    4D array (bonus!)


```python
import numpy as np
zerodim_array = np.array(42)
zerodim_array
```

```python
onedim_array = np.arange(5)
onedim_array
```

```python
twodim_array = np.arange(6).reshape(2,3)
twodim_array
```

```python
threedim_array = np.arange(8).reshape(2,2,2)
threedim_array
```

```python
fourdim_array = np.arange(8).reshape(2,2,1,2)
fourdim_array
```

## 3. Print the datatype, shape, size, and dimension of each of the above arrays
    Use .dtype, .shape, .size, and .ndim.


```python
print('1D Array:',onedim_array.dtype)
print('1D Array:',np.shape(onedim_array))
print('1D Array:',np.size(onedim_array))
print('1D Array:',np.ndim(onedim_array))
```

```python
print('2D Array:',twodim_array.dtype)
print('2D Array:',np.shape(twodim_array))
print('2D Array:',np.size(twodim_array))
print('2D Array:',np.ndim(twodim_array))
```

```python
print('3D Array:',threedim_array.dtype)
print('3D Array:',np.shape(threedim_array))
print('3D Array:',np.size(threedim_array))
print('3D Array:',np.ndim(threedim_array))
```

```python
print('4D Array:',fourdim_array.dtype)
print('4D Array:',np.shape(fourdim_array))
print('4D Array:',np.size(fourdim_array))
print('4D Array:',np.ndim(fourdim_array))
```
